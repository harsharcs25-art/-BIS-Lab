```python

import random
# Hospital staff
staff = ["A", "B", "C", "D", "E", "F"]
shifts = ["Morning", "Evening", "Night"]

POP_SIZE = 10
GENERATIONS = 50
MUTATION_RATE = 0.1

# Each chromosome represents staff assigned to 3 shifts
# Each shift requires 2 staff members
def create_schedule():
    return random.sample(staff, 6)

# Fitness: penalize repeated staff and uneven shift assignments
def fitness(chromosome):
    penalty = 0

    # Check each shift has 2 staff members
    for i in range(0, 6, 2):
        if len(set(chromosome[i:i+2])) < 2:
            penalty += 10

    return 1 / (1 + penalty)

# Select the better parent using tournament selection
def selection(population):
    candidates = random.sample(population, 3)
    return max(candidates, key=fitness)

# Crossover: combine two parents
def crossover(parent1, parent2):
    point = random.randint(1, 5)
    child = parent1[:point] + parent2[point:]

    # Repair duplicates to ensure every staff member appears once
    missing = [s for s in staff if s not in child]
    seen = set()

    for i in range(len(child)):
        if child[i] in seen:
            child[i] = missing.pop(0)
        else:
            seen.add(child[i])

    return child

# Mutation: swap two staff assignments
def mutation(chromosome):
    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(6), 2)
        chromosome[i], chromosome[j] = chromosome[j], chromosome[i]
    return chromosome

# Initialize population
population = [create_schedule() for _ in range(POP_SIZE)]

# Genetic Algorithm
for generation in range(GENERATIONS):
    new_population = []

    # Preserve the best solution (elitism)
    best = max(population, key=fitness)
    new_population.append(best[:])

    while len(new_population) < POP_SIZE:
        parent1 = selection(population)
        parent2 = selection(population)

        child = crossover(parent1, parent2)
        child = mutation(child)

        new_population.append(child)

    population = new_population

# Display best schedule
best = max(population, key=fitness)

print("Best Hospital Staff Schedule:")
for i, shift in enumerate(shifts):
    assigned = best[2*i:2*i+2]
    print(shift, ":", assigned)

print("Fitness:", fitness(best))
```

```python

import random

# Get inputs
num_staff = int(input("Enter number of staff: "))
num_shifts = int(input("Enter number of shifts: "))
pop_size = int(input("Enter population size: "))
crossover_rate = float(input("Enter crossover rate (0-1): "))
mutation_rate = float(input("Enter mutation rate (0-1): "))
generations = int(input("Enter number of generations: "))

# Each chromosome assigns a staff member to each shift
def create_schedule():
    return [random.randrange(num_staff) for _ in range(num_shifts)]

# Lower penalty means a better schedule
def fitness(schedule):
    # Penalize repeated staff assignments across shifts
    conflicts = len(schedule) - len(set(schedule))
    return 1 / (1 + conflicts)

# Initialize population
population = [create_schedule() for _ in range(pop_size)]

best_schedule = None
best_fitness = -1

for gen in range(generations):
    population.sort(key=fitness, reverse=True)

    if fitness(population[0]) > best_fitness:
        best_schedule = population[0][:]
        best_fitness = fitness(best_schedule)

    new_population = [population[0][:]]  # Preserve best solution

    while len(new_population) < pop_size:
        # Selection
        parent1 = random.choice(population[:max(1, pop_size // 2)])
        parent2 = random.choice(population[:max(1, pop_size // 2)])

        child1 = parent1[:]
        child2 = parent2[:]

        # Crossover
        if num_shifts > 1 and random.random() < crossover_rate:
            point = random.randint(1, num_shifts - 1)
            child1 = parent1[:point] + parent2[point:]
            child2 = parent2[:point] + parent1[point:]

        # Mutation
        for child in (child1, child2):
            for i in range(num_shifts):
                if random.random() < mutation_rate:
                    child[i] = random.randrange(num_staff)

            new_population.append(child)
            if len(new_population) >= pop_size:
                break

    population = new_population

# Output result
print("\nBest schedule found:")
for shift, staff in enumerate(best_schedule, start=1):
    print(f"Shift {shift}: Staff {staff + 1}")

print("Fitness:", best_fitness)
print("Repeated assignments:", num_shifts - len(set(best_schedule)))
```
